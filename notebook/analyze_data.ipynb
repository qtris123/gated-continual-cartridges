# Overlap Analysis: Eval Phase2 "Update" Questions vs Self-Study Data

Check how much the eval phase2 "update" category questions overlap with the self-study training data (data contamination/leakage check).

```python
import pandas as pd
import numpy as np
import re
from collections import Counter
import math

# Load both datasets
eval_df = pd.read_parquet("/home/phudishp/cartridges/experiments/continual_learning/data/eval/eval_phase2.parquet")
ss_df = pd.read_parquet("/home/phudishp/cartridges/outputs/2026-02-09-15-23-19-synthesize_amd/synthesize_amd_meta-llama/Llama-3.2-3B-Instruct_n65536-0/artifact/dataset_clean.parquet")

print(f"Eval phase2: {len(eval_df)} rows")
print(f"Self-study:  {len(ss_df)} rows")

# Filter eval to "update" category
eval_df["category"] = eval_df["metadata"].apply(lambda x: x.get("category", ""))
update_df = eval_df[eval_df["category"] == "update"].copy()
print(f"\nUpdate eval questions: {len(update_df)}")

# Extract user questions from both datasets
def extract_user_question(messages):
    """Extract the first user message content from a messages array."""
    for msg in messages:
        if msg["role"] == "user":
            return msg["content"]
    return ""

update_df["question"] = update_df["messages"].apply(extract_user_question)
ss_df["question"] = ss_df["messages"].apply(extract_user_question)

ss_questions = ss_df["question"].tolist()
update_questions = update_df["question"].tolist()

print(f"\nExtracted {len(update_questions)} update eval questions")
print(f"Extracted {len(ss_questions)} self-study questions")
```

```python
# Print all 20 update eval questions for manual inspection
print("=" * 80)
print("ALL 20 UPDATE EVAL QUESTIONS")
print("=" * 80)
for i, (idx, row) in enumerate(update_df.iterrows()):
    qid = row["metadata"].get("question_id", "?")
    diff = row["metadata"].get("difficulty", "?")
    print(f"\n[{i+1:2d}] ID={qid} | difficulty={diff}")
    print(f"    {row['question']}")
```

## 1. Exact Match Check

```python
# Exact match check
ss_question_set = set(ss_questions)
ss_question_set_lower = set(q.lower().strip() for q in ss_questions)

print("EXACT MATCH CHECK")
print("=" * 60)
exact_matches = 0
for i, q in enumerate(update_questions):
    if q in ss_question_set:
        print(f"  EXACT MATCH [{i+1}]: {q[:100]}")
        exact_matches += 1
    elif q.lower().strip() in ss_question_set_lower:
        print(f"  CASE-INSENSITIVE MATCH [{i+1}]: {q[:100]}")
        exact_matches += 1

if exact_matches == 0:
    print("  No exact matches found.")
print(f"\nTotal exact matches: {exact_matches} / {len(update_questions)}")
```

## 2. Keyword / Topic Overlap

```python
# Keyword/topic overlap analysis
# Extract key financial/AMD-specific entities from update questions
key_topics = [
    "revenue", "net income", "gross margin", "operating income", "operating expense",
    "eps", "earnings per share", "data center", "client", "gaming", "embedded",
    "xilinx", "pensando", "segment", "acquisition", "cash flow", "debt",
    "goodwill", "inventory", "r&d", "research and development",
    "total assets", "total liabilities", "stockholders' equity",
    "diluted", "shares outstanding", "capital expenditure", "capex",
    "fiscal year", "quarter", "yoy", "year-over-year",
]

print("KEYWORD/TOPIC OVERLAP")
print("=" * 60)

# For each topic, count how many update questions and self-study questions mention it
topic_overlap = []
for topic in key_topics:
    pattern = re.compile(re.escape(topic), re.IGNORECASE)
    n_update = sum(1 for q in update_questions if pattern.search(q))
    n_ss = sum(1 for q in ss_questions if pattern.search(q))
    if n_update > 0 or n_ss > 0:
        topic_overlap.append((topic, n_update, n_ss))

topic_df = pd.DataFrame(topic_overlap, columns=["Topic", "Update Eval (n=20)", "Self-Study (n=65536)"])
topic_df = topic_df.sort_values("Update Eval (n=20)", ascending=False)
print(topic_df.to_string(index=False))

print(f"\n--- Topics present in update eval questions ---")
for _, row in topic_df[topic_df["Update Eval (n=20)"] > 0].iterrows():
    pct_ss = row["Self-Study (n=65536)"] / len(ss_questions) * 100
    print(f"  '{row['Topic']}': {row['Update Eval (n=20)']} eval q / {row['Self-Study (n=65536)']} self-study q ({pct_ss:.2f}%)")
```

## 3. Semantic Similarity (TF-IDF Cosine Similarity)

```python
# TF-IDF implementation from scratch (sklearn not available)
import numpy as np
from collections import Counter
import re
import math

def tokenize(text):
    """Simple tokenizer: lowercase, split on non-alphanumeric."""
    return re.findall(r'[a-z0-9]+', text.lower())

def build_tfidf(documents):
    """Build TF-IDF matrix for a list of documents.
    Returns: (tfidf_matrix as list of sparse dicts, vocab, idf_dict)
    """
    # Build vocabulary and document frequencies
    doc_freq = Counter()
    tokenized = []
    for doc in documents:
        tokens = tokenize(doc)
        tokenized.append(tokens)
        unique_tokens = set(tokens)
        for t in unique_tokens:
            doc_freq[t] += 1

    n_docs = len(documents)
    # IDF with smoothing
    idf = {t: math.log((1 + n_docs) / (1 + df)) + 1 for t, df in doc_freq.items()}

    # Build TF-IDF as sparse dicts
    tfidf_docs = []
    for tokens in tokenized:
        tf = Counter(tokens)
        total = len(tokens) if tokens else 1
        tfidf = {}
        for t, count in tf.items():
            if t in idf:
                tfidf[t] = (count / total) * idf[t]
        # L2 normalize
        norm = math.sqrt(sum(v ** 2 for v in tfidf.values())) or 1.0
        tfidf = {t: v / norm for t, v in tfidf.items()}
        tfidf_docs.append(tfidf)

    return tfidf_docs, idf

def cosine_sim_sparse(a, b):
    """Cosine similarity between two sparse TF-IDF dicts (already L2-normalized)."""
    common_keys = set(a.keys()) & set(b.keys())
    return sum(a[k] * b[k] for k in common_keys)

print("Building TF-IDF for all questions...")
print("  (This may take a minute for 65K+ documents)")

# Combine all questions, eval first then self-study
all_questions = update_questions + ss_questions
tfidf_all, idf = build_tfidf(all_questions)

# Split back
n_update = len(update_questions)
tfidf_update = tfidf_all[:n_update]
tfidf_ss = tfidf_all[n_update:]

print(f"  TF-IDF built: {n_update} eval + {len(tfidf_ss)} self-study vectors")
print(f"  Vocabulary size: {len(idf)}")
```

```python
# Find top-5 most similar self-study questions for each update eval question
TOP_N = 5

print("=" * 100)
print(f"TOP-{TOP_N} MOST SIMILAR SELF-STUDY QUESTIONS FOR EACH UPDATE EVAL QUESTION")
print("=" * 100)

all_top_scores = []

for i, (uq, uv) in enumerate(zip(update_questions, tfidf_update)):
    # Compute similarity against all self-study questions
    sims = []
    for j, sv in enumerate(tfidf_ss):
        sim = cosine_sim_sparse(uv, sv)
        if sim > 0:
            sims.append((j, sim))

    # Sort by similarity descending
    sims.sort(key=lambda x: -x[1])
    top_n = sims[:TOP_N]
    all_top_scores.append([s for _, s in top_n])

    qid = update_df.iloc[i]["metadata"].get("question_id", "?")
    print(f"\n{'─' * 100}")
    print(f"[{i+1:2d}] EVAL Q (ID={qid}): {uq}")
    print(f"{'─' * 100}")
    for rank, (j, sim) in enumerate(top_n):
        print(f"  #{rank+1} (sim={sim:.4f}): {ss_questions[j]}")

print(f"\n{'=' * 100}")
```

## 4. Summary

```python
# Summary statistics
print("=" * 80)
print("OVERLAP ANALYSIS SUMMARY")
print("=" * 80)

# 1. Exact matches
print(f"\n1. EXACT MATCHES: {exact_matches} / {len(update_questions)}")

# 2. Similarity score statistics
top1_scores = [scores[0] if scores else 0.0 for scores in all_top_scores]
print(f"\n2. TF-IDF COSINE SIMILARITY (top-1 match per eval question):")
print(f"   Mean:   {np.mean(top1_scores):.4f}")
print(f"   Median: {np.median(top1_scores):.4f}")
print(f"   Min:    {np.min(top1_scores):.4f}")
print(f"   Max:    {np.max(top1_scores):.4f}")
print(f"   Std:    {np.std(top1_scores):.4f}")

# Thresholds
thresholds = [0.8, 0.7, 0.6, 0.5, 0.4, 0.3]
print(f"\n3. NUMBER OF EVAL QUESTIONS WITH TOP-1 SIMILARITY ABOVE THRESHOLD:")
for t in thresholds:
    count = sum(1 for s in top1_scores if s >= t)
    print(f"   >= {t:.1f}: {count:2d} / {len(update_questions)} ({count/len(update_questions)*100:.0f}%)")

# Per-question summary table
print(f"\n4. PER-QUESTION SUMMARY (top-1 similarity):")
print(f"{'#':>3}  {'ID':>4}  {'Top-1 Sim':>9}  {'Eval Question'}")
print(f"{'─'*3}  {'─'*4}  {'─'*9}  {'─'*60}")
for i, (q, scores) in enumerate(zip(update_questions, all_top_scores)):
    qid = update_df.iloc[i]["metadata"].get("question_id", "?")
    s = scores[0] if scores else 0.0
    flag = " ⚠️ HIGH" if s >= 0.7 else ""
    print(f"{i+1:3d}  {qid:>4}  {s:9.4f}  {q[:60]}{flag}")
```
